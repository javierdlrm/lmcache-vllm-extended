```python
import os
import pandas as pd

csv_dir = "reports/rag_benchmark_final/"
csv_files = [f for f in os.listdir(csv_dir) if f.endswith('.csv')]

stats = []

for file in csv_files:
    df = pd.read_csv(os.path.join(csv_dir, file))
    row = {'file': file}
    for col in ['rag_latency', 'rag_encode_latency', 'rag_search_latency']:
        if col in df.columns:
            values = df[col].dropna()
            row[f'{col}_avg'] = values.mean().round(5)
            row[f'{col}_median'] = values.median().round(5)
            row[f'{col}_p75'] = values.quantile(0.75).round(5)
            row[f'{col}_p99'] = values.quantile(0.99).round(5)
        else:
            row[f'{col}_avg'] = None
            row[f'{col}_median'] = None
            row[f'{col}_p50'] = None
            row[f'{col}_p75'] = None
            row[f'{col}_p90'] = None
            row[f'{col}_p95'] = None
            row[f'{col}_p99'] = None
    stats.append(row)

stats_df = pd.DataFrame(stats)

# Sort by the numeric value in the file name
stats_df_sorted = stats_df.copy()
stats_df_sorted["file_num"] = stats_df_sorted["file"].str.extract(r'rag_benchmark_(\d+)\.csv').astype(int)
stats_df_sorted = stats_df_sorted.sort_values(by="file_num").reset_index(drop=True)
stats_df_sorted = stats_df_sorted.drop(columns=["file_num"])

stats_df_sorted


```
